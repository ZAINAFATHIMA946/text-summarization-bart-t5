```python
import requests
from bs4 import BeautifulSoup

url = "https://rvu.edu.in/rvu-at-a-glance/"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Extract paragraphs
paragraphs = soup.find_all("p")
text = " ".join([p.get_text() for p in paragraphs])

print(text[:1000])  # preview first 1000 chars
```

```python
import re

clean_text = re.sub(r'\s+', ' ', text)  # remove extra spaces
clean_text = clean_text.strip()

print(clean_text[:1000])
```

```python
from transformers import BartTokenizer, BartForConditionalGeneration

model_name = "facebook/bart-large-cnn"
tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)

text = clean_text

# BART models typically handle longer sequences, up to 1024 tokens
inputs = tokenizer([text], max_length=1024, return_tensors="pt", truncation=True)

# Generate summary
summary_ids = model.generate(
    inputs["input_ids"],
    num_beams=4, # Use beam search for better quality
    max_length=150, # Max length of the generated summary
    min_length=30,  # Min length of the generated summary
    early_stopping=True
)

summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print(summary_text)
```

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

model_name = "t5-base"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)

text = clean_text
inputs = tokenizer(text, return_tensors="pt", max_length=512, truncation=True)

outputs = model.generate(inputs["input_ids"], max_length=150)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
# ACTIVITY 1: Load the BART summarization model and tokenizer explicitly
from transformers import BartTokenizer, BartForConditionalGeneration

model_name = "facebook/bart-large-cnn"
tokenizer = BartTokenizer.from_pretrained(model_name)
bart_model = BartForConditionalGeneration.from_pretrained(model_name)

print("BART model and tokenizer loaded successfully!")
```

```python
# Generate a summary using BART (manually tokenizing and generating)

# Map the news article text (stored in clean_text)
news_article = clean_text

# Tokenize the input text
inputs = tokenizer(
    [news_article],
    max_length=1024,
    return_tensors="pt",
    truncation=True
)

# Generate summary IDs using the model
summary_ids = bart_model.generate(
    inputs["input_ids"],
    max_length=100,
    min_length=25,
    num_beams=4,
    early_stopping=True
)

# Decode the summary token IDs back into readable text
summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("ORIGINAL ARTICLE:\n")
print(news_article[:1000] + "...") # Preview the first 1000 characters

print("\nBART GENERATED SUMMARY:\n")
print(summary_text)
```

```python
# Compare the original article and the generated summary

original_words = len(news_article.split())
summary_words = len(summary_text.split())

print("Original Article Word Count:", original_words)
print("Summary Word Count:", summary_words)

if original_words > 0:
    reduction = (1 - summary_words / original_words) * 100
    print("Word Count Reduction:", round(reduction, 2), "%")
```

```python
# ACTIVITY 2: Load the T5 summarization model
from transformers import T5Tokenizer, T5ForConditionalGeneration

t5_model_name = "google-t5/t5-base"
t5_tokenizer = T5Tokenizer.from_pretrained(t5_model_name)
t5_model = T5ForConditionalGeneration.from_pretrained(t5_model_name)

print("T5 model loaded successfully!")
```

```python
# Generate a summary using T5

# T5 requires the task prefix 'summarize: ' before the text
t5_input_text = "summarize: " + news_article

# Tokenize the input text
inputs = t5_tokenizer(
    [t5_input_text],
    max_length=512,
    return_tensors="pt",
    truncation=True
)

# Generate summary IDs using the model
summary_ids = t5_model.generate(
    inputs["input_ids"],
    max_length=100,
    min_length=25,
    num_beams=4,
    early_stopping=True
)

# Decode the summary token IDs back into readable text
t5_summary_text = t5_tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("T5 GENERATED SUMMARY:\n")
print(t5_summary_text)
```

```python
# Compare the BART and T5 summaries

bart_word_count = len(summary_text.split())
t5_word_count = len(t5_summary_text.split())

print("=" * 60)
print("BART SUMMARY")
print("=" * 60)
print(summary_text)

print("\nWord Count:", bart_word_count)

print("\n" + "=" * 60)
print("T5 SUMMARY")
print("=" * 60)
print(t5_summary_text)

print("\nWord Count:", t5_word_count)
```

```python
# Comparison table

import pandas as pd

comparison = pd.DataFrame({
    "Evaluation Criteria": [
        "Model",
        "Summary Word Count",
        "Main Topic",
        "Readability",
        "Information Preservation"
    ],
    "BART": [
        "facebook/bart-large-cnn",
        bart_word_count,
        "AI in healthcare",
        "Evaluate generated output",
        "Check key points retained"
    ],
    "T5": [
        "google-t5/t5-base",
        t5_word_count,
        "AI in healthcare",
        "Evaluate generated output",
        "Check key points retained"
    ]
})

display(comparison)
```

Observation:

Both BART and T5 can generate abstractive summaries. Their outputs may differ in wording, length, readability, and the amount of information preserved.

ACTIVITY 3: Effect of Summary Length
Aim

To investigate how changing the max_length and min_length parameters affects the generated summary.

In the Hugging Face summarization pipeline:

max_length controls the maximum number of generated tokens.
min_length specifies the minimum desired number of generated tokens.

These values refer to tokens, not words.

```python
# ACTIVITY 3: Effect of Summary Length

length_settings = [
    {"max_length": 50, "min_length": 10},
    {"max_length": 100, "min_length": 25},
    {"max_length": 150, "min_length": 50}
]

length_results = []

for settings in length_settings:
    # Tokenize the input text
    inputs = tokenizer(
        [news_article],
        max_length=1024,
        return_tensors="pt",
        truncation=True
    )

    # Generate summary using the loaded BART model directly
    summary_ids = bart_model.generate(
        inputs["input_ids"],
        max_length=settings["max_length"],
        min_length=settings["min_length"],
        num_beams=4,
        early_stopping=True
    )

    # Decode output
    generated_summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

    length_results.append({
        "Max Length": settings["max_length"],
        "Min Length": settings["min_length"],
        "Summary": generated_summary,
        "Word Count": len(generated_summary.split())
    })

    print("\n" + "=" * 60)
    print("Maximum Length:", settings["max_length"])
    print("Minimum Length:", settings["min_length"])
    print("\nGenerated Summary:\n")
    print(generated_summary)
    print("\nWord Count:", len(generated_summary.split()))
```

```python
# Compare summaries generated with different length settings

length_comparison = pd.DataFrame(length_results)

display(length_comparison)
```

Observation:

Changing the generation length parameters affects the length and potentially the completeness of the summary. The best setting depends on the document and the intended use.

ACTIVITY 4: Summarization of Research Abstracts
Aim

To apply the summarization model to research abstracts from different domains and evaluate whether the summaries preserve the research problem, methodology, and key findings.

We'll use two illustrative research abstracts:

Artificial Intelligence in healthcare.
Machine Learning in education.

These are sample abstracts created for this laboratory exercise, not abstracts taken from published research papers.

```python
# ACTIVITY 4: Research Abstract Summarization

healthcare_abstract = """
Research Problem:
Healthcare professionals face challenges in identifying diseases
early because medical data can be large and complex.

Methodology:
This study explores the use of machine learning techniques to analyze
medical images and patient records. The proposed approach examines
patterns in the data to support disease prediction.

Key Findings:
The approach demonstrates the potential of machine learning to assist
in medical image analysis and early risk identification. However,
reliable deployment requires careful evaluation, representative data,
and human supervision.

Conclusion:
Machine learning can support clinical decision-making, but it should
be used as an aid to healthcare professionals rather than as a
replacement for their judgment.
"""

education_abstract = """
Research Problem:
Students have different learning needs, and traditional teaching
methods may not provide personalized feedback to every student.

Methodology:
This study investigates an artificial intelligence-based learning
system that analyzes student performance and recommends learning
materials. The system uses performance patterns to identify topics
that may require additional practice.

Key Findings:
Personalized recommendations may help students focus on difficult
topics and provide teachers with useful information about learning
progress. The effectiveness of such systems depends on the quality
of the data and the suitability of the recommendations.

Conclusion:
AI-based learning tools can support personalized education when used
alongside appropriate teaching practices and teacher supervision.
"""

print("Research abstracts prepared successfully!")
```

```python
# Summarize the healthcare research abstract using the loaded BART model directly

# Tokenize the input text
inputs = tokenizer(
    [healthcare_abstract],
    max_length=1024,
    return_tensors="pt",
    truncation=True
)

# Generate summary IDs
summary_ids = bart_model.generate(
    inputs["input_ids"],
    max_length=100,
    min_length=25,
    num_beams=4,
    early_stopping=True
)

# Decode output
healthcare_summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("HEALTHCARE RESEARCH ABSTRACT:\n")
print(healthcare_abstract)

print("\nGENERATED SUMMARY:\n")
print(healthcare_summary)
```

```python
# Summarize the education research abstract using the loaded BART model directly

# Tokenize the input text
inputs = tokenizer(
    [education_abstract],
    max_length=1024,
    return_tensors="pt",
    truncation=True
)

# Generate summary IDs
summary_ids = bart_model.generate(
    inputs["input_ids"],
    max_length=100,
    min_length=25,
    num_beams=4,
    early_stopping=True
)

# Decode output
education_summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("EDUCATION RESEARCH ABSTRACT:\n")
print(education_abstract)

print("\nGENERATED SUMMARY:\n")
print(education_summary)
```

```python
# Evaluate preservation of key research information

evaluation = pd.DataFrame([
    {
        "Domain": "Healthcare",
        "Research Problem Preserved": (
            "Yes" if "disease" in healthcare_summary.lower()
            or "medical" in healthcare_summary.lower()
            else "Review manually"
        ),
        "Methodology Preserved": (
            "Yes" if "machine learning" in healthcare_summary.lower()
            or "data" in healthcare_summary.lower()
            else "Review manually"
        ),
        "Findings Preserved": (
            "Yes" if "assist" in healthcare_summary.lower()
            or "support" in healthcare_summary.lower()
            else "Review manually"
        ),
        "Generated Summary": healthcare_summary
    },
    {
        "Domain": "Education",
        "Research Problem Preserved": (
            "Yes" if "student" in education_summary.lower()
            or "learning" in education_summary.lower()
            else "Review manually"
        ),
        "Methodology Preserved": (
            "Yes" if "system" in education_summary.lower()
            or "performance" in education_summary.lower()
            else "Review manually"
        ),
        "Findings Preserved": (
            "Yes" if "support" in education_summary.lower()
            or "recommend" in education_summary.lower()
            else "Review manually"
        ),
        "Generated Summary": education_summary
    }
])

display(evaluation)
```

Observation:

The summarization model can condense research abstracts from different domains. However, the generated summaries must be checked to ensure that they preserve the main problem, the research methodology, and the important findings without introducing unsupported claims.
